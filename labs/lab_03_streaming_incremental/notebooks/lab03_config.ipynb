## Read and normalize parameter values

```python
catalog = dbutils.widgets.get("catalog").strip()
schema = dbutils.widgets.get("schema").strip()
volume_name = dbutils.widgets.get("volume_name").strip()

storage_account = dbutils.widgets.get("storage_account").strip()
container = dbutils.widgets.get("container").strip()
external_volume_dir = (
    dbutils.widgets.get("external_volume_dir")
    .strip()
    .strip("/")
)

source_dir = (
    dbutils.widgets.get("source_dir")
    .strip()
    .strip("/")
)

source_file_name = (
    dbutils.widgets.get("source_file_name")
    .strip()
)

landing_dir = (
    dbutils.widgets.get("landing_dir")
    .strip()
    .strip("/")
)

staging_dir = (
    dbutils.widgets.get("staging_dir")
    .strip()
    .strip("/")
)

autoloader_schema_dir = (
    dbutils.widgets.get("autoloader_schema_dir")
    .strip()
    .strip("/")
)

autoloader_checkpoint_dir = (
    dbutils.widgets.get("autoloader_checkpoint_dir")
    .strip()
    .strip("/")
)

eventhub_bronze_checkpoint_dir = (
    dbutils.widgets.get("eventhub_bronze_checkpoint_dir")
    .strip()
    .strip("/")
)

eventhub_silver_checkpoint_dir = (
    dbutils.widgets.get("eventhub_silver_checkpoint_dir")
    .strip()
    .strip("/")
)

file_bronze_table_name = (
    dbutils.widgets.get("file_bronze_table")
    .strip()
)

file_silver_table_name = (
    dbutils.widgets.get("file_silver_table")
    .strip()
)

eventhub_bronze_table_name = (
    dbutils.widgets.get("eventhub_bronze_table")
    .strip()
)

eventhub_silver_table_name = (
    dbutils.widgets.get("eventhub_silver_table")
    .strip()
)

trigger_type = (
    dbutils.widgets.get("trigger_type")
    .strip()
)

max_files_per_trigger = int(
    dbutils.widgets.get("max_files_per_trigger")
    .strip()
)

schema_evolution_mode = (
    dbutils.widgets.get("schema_evolution_mode")
    .strip()
)

rescued_data_column = (
    dbutils.widgets.get("rescued_data_column")
    .strip()
)

create_folder_markers = (
    dbutils.widgets.get("create_folder_markers")
    .strip()
    .lower()
    == "true"
)

secret_scope = (
    dbutils.widgets.get("secret_scope")
    .strip()
)

producer_id = (
    dbutils.widgets.get("producer_id")
    .strip()
)

eventhub_name = (
    dbutils.widgets.get("eventhub_name")
    .strip()
)

eventhub_consumer_group = (
    dbutils.widgets.get("eventhub_consumer_group")
    .strip()
)

eventhub_secret_key = (
    dbutils.widgets.get("eventhub_secret_key")
    .strip()
)
```

## Validate parameters

```python
required_values = {
    "catalog": catalog,
    "schema": schema,
    "volume_name": volume_name,
    "storage_account": storage_account,
    "container": container,
    "external_volume_dir": external_volume_dir,
    "source_dir": source_dir,
    "source_file_name": source_file_name,
    "landing_dir": landing_dir,
    "staging_dir": staging_dir,
    "autoloader_schema_dir": autoloader_schema_dir,
    "autoloader_checkpoint_dir": autoloader_checkpoint_dir,
    "eventhub_bronze_checkpoint_dir": (
        eventhub_bronze_checkpoint_dir
    ),
    "eventhub_silver_checkpoint_dir": (
        eventhub_silver_checkpoint_dir
    ),
    "file_bronze_table": file_bronze_table_name,
    "file_silver_table": file_silver_table_name,
    "eventhub_bronze_table": eventhub_bronze_table_name,
    "eventhub_silver_table": eventhub_silver_table_name,
    "secret_scope": secret_scope,
    "producer_id": producer_id,
    "eventhub_name": eventhub_name,
    "eventhub_consumer_group": eventhub_consumer_group,
    "eventhub_secret_key": eventhub_secret_key,
}

missing_values = [
    name
    for name, value in required_values.items()
    if not value
]

if missing_values:
    raise ValueError(
        f"Missing required parameters: {missing_values}"
    )

if max_files_per_trigger <= 0:
    raise ValueError(
        "max_files_per_trigger must be greater than zero"
    )

allowed_trigger_types = {
    "availableNow",
    "once",
    "processingTime",
}

if trigger_type not in allowed_trigger_types:
    raise ValueError(
        f"Unsupported trigger type: {trigger_type}"
    )

allowed_schema_modes = {
    "addNewColumns",
    "rescue",
    "failOnNewColumns",
    "none",
}

if schema_evolution_mode not in allowed_schema_modes:
    raise ValueError(
        f"Unsupported schema evolution mode: "
        f"{schema_evolution_mode}"
    )

if not rescued_data_column:
    raise ValueError(
        "rescued_data_column cannot be empty"
    )

if not eventhub_secret_key:
    raise ValueError(
        "eventhub_secret_key cannot be empty"
    )

print("Configuration validation passed.")
```
