### Reconcile Gold order-item totals

**Purpose:** Calculate the Gold fact metrics used by the reconciliation contract.

**Inputs:** `dbr_dev.parvinbadov.gold_fact_order_items`.

**Outputs:** Row count, distinct orders, and rounded monetary totals at the overall fact-table grain.

**Why it matters:** The result makes metric drift visible before publication.

```sql
-- Purpose:
-- Calculate the main Gold fact-table metrics used for reconciliation.

SELECT
    COUNT(*) AS actual_rows,
    COUNT(DISTINCT order_id) AS actual_orders,

    ROUND(
        SUM(price),
        2
    ) AS actual_total_price,

    ROUND(
        SUM(freight_value),
        2
    ) AS actual_total_freight,

    ROUND(
        SUM(item_total_value),
        2
    ) AS actual_total_value

FROM
    dbr_dev.parvinbadalov.gold_fact_order_items;
```
