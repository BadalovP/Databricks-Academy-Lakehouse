### Inspect the latest pipeline alert status

**Purpose:** Read the single latest alert-status record used by the Databricks SQL Alert.

**Inputs:** `dbr_dev.parvinbadov.olist_pipeline_alert_status`.

**Outputs:** Alert value, validation statuses, message, and validation timestamp.

**Why it matters:** The alert condition surfaces failed reconciliation or integrity checks to operators.

```sql
-- Purpose:
-- Return the latest numeric pipeline status for the Databricks SQL Alert.
--
-- Alert condition:
-- FIRST_ROW(alert_value) > 0

SELECT
    alert_value,
    overall_status,
    integrity_status,
    alert_message,
    latest_validation_at
FROM
    dbr_dev.parvinbadalov.olist_pipeline_alert_status;
```
