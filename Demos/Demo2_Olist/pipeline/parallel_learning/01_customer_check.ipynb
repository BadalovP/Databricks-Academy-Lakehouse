### Validate the Gold customer dimension

**Purpose:** Confirm that the customer dimension contains usable records.

**Inputs:** `dbr_dev.parvinbadov.gold_dim_customer`.

**Outputs:** Customer row count and PASS task values for the summary branch.

**Why it matters:** An empty customer dimension would make governed customer analysis incomplete.

```python
# Purpose: validate the Gold customer table in the first parallel branch.

customer_table = "dbr_dev.parvinbadalov.gold_dim_customer"

customer_rows = spark.table(customer_table).count()

if customer_rows <= 0:
    raise RuntimeError(
        f"Customer table is empty: {customer_table}"
    )

dbutils.jobs.taskValues.set(
    key="customer_rows",
    value=customer_rows
)

dbutils.jobs.taskValues.set(
    key="customer_status",
    value="PASS"
)

print("Customer table:", customer_table)
print("Customer rows:", customer_rows)
print("Customer validation: PASS")
```
