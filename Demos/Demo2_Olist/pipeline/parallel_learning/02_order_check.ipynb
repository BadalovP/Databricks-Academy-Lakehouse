```python
# Purpose: validate the Gold order-item table in the second parallel branch.

from pyspark.sql import functions as F

order_table = "dbr_dev.parvinbadalov.gold_fact_order_items"

order_result = (
    spark.table(order_table)
    .agg(
        F.count("*").alias("order_item_rows"),
        F.countDistinct("order_id").alias("distinct_orders")
    )
    .first()
)

order_item_rows = order_result["order_item_rows"]
distinct_orders = order_result["distinct_orders"]

if order_item_rows <= 0:
    raise RuntimeError(
        f"Order-item table is empty: {order_table}"
    )

dbutils.jobs.taskValues.set(
    key="order_item_rows",
    value=order_item_rows
)

dbutils.jobs.taskValues.set(
    key="distinct_orders",
    value=distinct_orders
)

dbutils.jobs.taskValues.set(
    key="order_status",
    value="PASS"
)

print("Order table:", order_table)
print("Order-item rows:", order_item_rows)
print("Distinct orders:", distinct_orders)
print("Order validation: PASS")
```
